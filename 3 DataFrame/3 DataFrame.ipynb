```python
import numpy as np 
import pandas as pd
```

```python
# 1 What us DataFrame
# A DataFrame is a two-dimensional labeled table with rows and columns.
df = pd.DataFrame({
    "Name":["Jenish","Kriya"],
    "Marks":[85,90]
})
print(df)
```

```python
# 2 Creating DataFrame
# Use pd.DataFrame() to create a DataFrame.
data = [[10,20],
        [30,40]]
df = pd.DataFrame(data)
print(df)
```

```python
# 3 DataFrame from Dictionary
# Dictionary keys become column names and dictionary values become column data.
data = {
    "Name":["Jenish","Kriya","Parth"],
    "Marks":[80,90,75]
}
df = pd.DataFrame(data)
print(df)
```

```python
# 4 DataFrame from List
# A list of lists can be converted into a DataFrame.
data = [
    ["Jenish",80],
    ["Kriya",90],
    ["Parth",75]
]
df = pd.DataFrame(data,columns=["Name","Marks"])
print(df)
```

```python
# 5 DataFrame from NumPy Array
# A 2D NumPy array can be converted into a DataFrame.
arr = np.array([
    [10,20],
    [30,40],
    [50,60]
])

df = pd.DataFrame(arr,columns=["A","B"])
print(df)
```

```python
# 6 DataFrame from List of Dictionaries
# Each dictionary represents one row. Keys become columns.
data = [
    {"Name":"Jenish","Marks":90},
    {"Name":"Kriya","Marks":80},
    {"Name":"Parth","Marks":10},
]

df = pd.DataFrame(data)
print(df)
```

```python
# 7 Custoom Index 
# You can give your own row labels using index=.
data = {
    "Name":["Jenish","Kriya"],
    "Marks":["80,90"]
}

df = pd.DataFrame(
    data,
    index = ["Student1","Student2"]
)
print(df)
```

```python
# 8 Columns Names
# Use columns= to give meaningful names to columns.
data = [[10,20],[30,40]]
df = pd.DataFrame(
    data,columns=["Math","Science"]
)
print(df)
```

```python
# 9 Index 
# The index property shows row labels 
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"],
    "Marks":[80,90,75]
})
print(df.index)
```

```python
# 10 columns
# The columns property shows all column names.
print(df.columns)
```

```python
# 11 Values
# The values property returns the data without row and column labels.
print(df.values)
```

```python
# 12 shpae
# shape returns the number of rows and columns
print(df.shape)
```

```python
# 13 Size 
# size returns the total number of cells.
print(df.size)
```

```python
# 14 ndim
# ndim returns the number of dimensions.
print(df.ndim)
```

```python
# 15 dtypes
# dtypes shows the data type of each column.
print(df.dtypes)
```

```python
# 16 info
#  gives a quick structural summary: rows, columns, non-null values, data types and memory information.
df.info()
```

```python
# 17 describe
# describe() gives statistical information for numeric columns
print(df.describe())
```

```python
# 18 head
#  displays the first 5 rows by default. You can specify a number.
print(df.head(2))
```

```python
# 19 tail
#  displays the last 5 rows by default.
print(df.tail(2))
```

```python
# 20 sample
#  returns random rows from the DataFrame.
print(df.sample(2,random_state=1))
```

```python
# Final Practice Program

data = {
 "Name": ["Amit", "Neha", "Raj", "Priya"],
 "Age": [22, 21, 23, 22],
 "Marks": [80, 92, 75, 88]
}
df = pd.DataFrame(data)
print(df)

print("Shape:", df.shape)
print("Size:", df.size)
print("Dimensions:", df.ndim)
print("Columns:", df.columns)

print("Data types:")
print(df.dtypes)

print("First 2:")
print(df.head(2))

print("Last 2:")
print(df.tail(2))

print("Random 2:")
print(df.sample(2, random_state=1))

print("Statistics:")
print(df.describe())
```
