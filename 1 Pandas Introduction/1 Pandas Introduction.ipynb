```python
import numpy as np
import pandas as pd
```

```python
# 1 What is Pandas
# Pandas is a Python library used for working with data.
# It makes data easy to read, clean, analyze, filter, and organize.

# Pandas mainly works with two objects:
# 1. Series — one-dimensional data.
# 2. DataFrame — two-dimensional table-like data.
data = [10,20,30]
s = pd.Series(data)
print(s)
```

```python
# 2 Why use Pandas
# Real-world data is often stored in tables. Pandas helps us read CSV/Excel files, inspect data, select rows
# and columns, filter data, handle missing values, remove duplicates, group data, and merge datasets.
data = {
    "Name": ["Jenish","Kriya","Parth"],
    "Marks": [85,72,63]
}
df = pd.DataFrame(data)
print(df)
```

```python
# 3 Pandas vs Numpy 
# NumPy is mainly designed for numerical arrays and mathematical operations. Pandas is mainly designed
# for labeled and tabular data. Pandas works closely with NumPy.

a = np.array([10,20,30])
b = pd.Series([10,20,30])
print(a)
print(b)
```
