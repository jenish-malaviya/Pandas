```python
import pandas as pd
```

```python
# 1 What is Aggregation
# Aggregation means combining many values into one summary value, such as sum, mean, minimum, maximum, or
# count
df = pd.DataFrame({
 "Department": ["IT", "IT", "HR", "HR"],
 "Salary": [50000, 60000, 40000, 45000]
})
print(df["Salary"].sum())
print(df["Salary"].mean())
```

```python
# 2 agg
# applies one or more aggregation functions to a Series or DataFrame.
df = pd.DataFrame({
 "Salary": [50000, 60000, 40000, 45000]
})
result = df["Salary"].agg("mean")
print(result)
```

```python
# 3 Multiple Functions
# You can pass a list of functions to agg().
result = df["Salary"].agg(["sum", "mean", "min", "max"])
print(result)
```

```python
# 4 Custom Aggregation
# A custom aggregation is your own function. Here, we calculate the salary range.
def salary_range(x):
    return x.max() - x.min()

result = df["Salary"].agg(salary_range)
print(result)
```

```python
# 5 Column-wise Aggregation
# With axis=0, Pandas works down each column. This is the default for DataFrame aggregation
f = pd.DataFrame({
 "Salary": [50000, 60000, 40000],
 "Bonus": [5000, 6000, 4000]
})
result = df.agg("sum",axis=0)
print(result)
```

```python
# 6 axis=0
# axis=0 means aggregate vertically, column by column.
df = pd.DataFrame({
 "Salary": [50000, 60000, 40000],
 "Bonus": [5000, 6000, 4000]
})

result = df.agg("mean",axis=0)
print(result)
```

```python
# 7 Row-wia=se aggregation
# With axis=1, Pandas works across columns for each row.
result = df.agg("sum",axis=1)
print(result)
```

```python
# 8 axis = 1
# axis=1 means aggregate horizontally, row by row
result = df.agg("mean",axis=1)
print(result)
```

```python
# 9 Named Aggregation
# Named aggregation lets you choose the output column name and aggregation function.
df = pd.DataFrame({
 "Department": ["IT", "IT", "HR", "HR"],
 "Salary": [50000, 60000, 40000, 45000]
})

result = df.groupby("Department").agg(
    Average_Salary = ("Salary","mean")
)
print(result)
```

```python
# 10 Multiple Named Aggregations
result = df.groupby("Department").agg(
     Total_Salary=("Salary", "sum"),
     Average_Salary=("Salary", "mean"),
     Minimum_Salary=("Salary", "min"),
     Maximum_Salary=("Salary", "max"),
     Employee_Count=("Salary", "count")
)
print(result)
```

```python
# 11 Grouped Aggregation
# GroupBy combined with agg() is one of the most useful Pandas patterns for data analysis.
result = df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "min", "max", "count"]
)
print(result)
```

```python
# 12  Multiple Columns with agg()
# You can apply different aggregation functions to different columns.
df = pd.DataFrame({
    "Sales":[100,200,300],
    "Profit":[20,50,80]
})
result = df.agg({
    "Sales":["sum","mean"],
    "Profit":["min","max"]
})
print(result)
```

```python
# 13 Custom Aggregation on a DataFrame
# You can create a custom function and apply it to columns
def data_range(x):
    return x.max() - x.min()
    
result = df.agg(data_range)
print(result)
```

```python
# Quick Revision

# agg() Apply one or more aggregation functions
# Multiple functions Use a list such as ['sum', 'mean', 'max']
# Custom aggregation Use your own Python function
# axis=0 Work down columns / column-wise
# axis=1 Work across rows / row-wise
# Named aggregation Choose meaningful output names
# Grouped aggregation groupby() + agg()
```
