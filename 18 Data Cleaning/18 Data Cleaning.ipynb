```python
import numpy as np
import pandas as pd
```

```python
# 1 Missing Values
# Use isna() to find missing values and fillna() to replace them

df = pd.DataFrame({
    "Name":["A","B","C"],
    "Age":[20,None,30]
})

# isna is a find missing value miss -> True , notmiss -> False
print(df.isna())
# fillna find messing value and fill the value 
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
# 3 Wrong Data Types
# Convert numbers stored as text into numeric data with pd.to_numeric()
df = pd.DataFrame({
    "Age":["20","25","30"]
})
print(df.dtypes)

df["Age"] = pd.to_numeric(df["Age"])
print(df.dtypes)
```

```python
# 4 incoorect values
# Check a business rule and replace invalid values. Here, negative age is invalid.
df = pd.DataFrame({
    "Name":["A","B","C"],
    "Age":[20,-5,30]
})
# .loc = "Go to these rows and these columns."
# df.loc[rows, columns]
df.loc[df["Age"] < 0, "Age"] = np.nan # Not a Number.
print(df)
```

```python
# 5 Outliers
# The IQR method finds unusually high or low values.

df = pd.DataFrame({
    "Salary":[30000,32000,35000,34000,36000,200000]
})

Q1 = df["Salary"].quantile(.25)
Q3 = df["Salary"].quantile(.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower:",lower)
print("Upper:",upper)

print(df[(df["Salary"] < lower) | (df["Salary"] > upper)])
```

```python
# 6 Text Cleaning
# Remove extra spaces and standardize text case.

df = pd.DataFrame({
    "City":["   Pune","Mumbai   ","   Delhi"]
})
df["City"] = df["City"].str.strip().str.lower()
print(df)
```

```python
# 7 Numeric Cleaning
# Remove currency symbols and commas, then convert text to integers.
df = pd.DataFrame({
    "Sales":["₹1,200","₹2,500","₹3000"]
})

df["Sales"] = (df["Sales"]
              .str.replace("₹","",regex=False) # regex safe and clear choice.
              .str.replace(",","",regex=False)
              .astype(int))
print(df)
print(df["Sales"].dtype)
```

```python
# 8 Data Cleaning
# Convert date text into Pandas datetime values.
df = pd.DataFrame({
    "Date":["2026-01-10","2026-02-15","2026-03-20"]
})

df["Date"] = pd.to_datetime(df["Date"])
print(df)
print(df["Date"].dtype)
```

```python
# 9 Standardizing values
# Convert different spellings or abbreviations into one consistent value.
df = pd.DataFrame({
    "Gender":["Male","M","male","Female","F"]
})
df["Gender"] = df["Gender"].replace({
    "M":"Male",
    "male":"Male",
    "F":"Female"
})

print(df)
```

```python
# 10 Renaming Columns
# Use rename() to make column names clear.
df = pd.DataFrame({
    "nm": ["A","B"],
    "ag": [20,25]
})
df = df.rename(columns={
    "nm": "Name",
    "ag": "Age"
})
print(df)
```

```python
# 11 Removing Unwanted Columns
# Use drop(columns=...) to remove unnecessary columns.
df = pd.DataFrame({
    "Name":["A","B"],
    "Age":[20,25],
    "Temp":[100,200]
})

df = df.drop(columns=["Temp"])
print(df)
```

```python
# 12 Removing Unwanted Rows
# Use boolean filtering to keep only rows that satisfy a condition.
df = pd.DataFrame({
    "Name":["A","B","C"],
    "Age":[20,15,25]
})
df = df[df["Age"] >= 18]
print(df)
```

```python
# 13 Data Validation
# Validate data against an expected rule. Here, valid age must be 18–60.
df = pd.DataFrame({
    "Name":["A","B","C"],
    "Age":[20,70,25]
})
valid = df["Age"].between(18,60)
print(valid)
```
