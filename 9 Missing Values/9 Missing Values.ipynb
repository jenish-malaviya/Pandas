```python
import pandas as pd
```

```python
# 1 Missing Values
# Missing data means a value is unavailable or was not recorded.
df = pd.DataFrame({"Name":["Jenish",
                           "Kriya",
                           "Parth"], 
                   "Age":[22,
                          None,
                          30]})
print(df)
```

```python
# 2 NaN
# NaN means Not a Number and is commonly used for missing numeric values.
df = pd.DataFrame({"Age":[22,float("nan"),30]})
print(df)
print(df["Age"].isna())
```

```python
# 3 None
# None is Python's null value and pandas recognizes it as missing.
df = pd.DataFrame({"Name":["Jenish",None,"Parth"]})
print(df)
print(df["Name"].isna())
```

```python
# 4 pd.NA
# pd.NA is pandas' dedicated missing-value marker and works with nullable dtypes.
df = pd.DataFrame({"Name":pd.Series(["Jenish",pd.NA,"Parth"],dtype="string")})
print(df)
print(df["Name"].isna())
```

```python
# 5 Detect Missing Values
# isna() checks every cell and returns True where data is missing.
df = pd.DataFrame({
    "Name":["Jenish","Kriya",None],
    "Age":[22,25,None],
    "Salary":[30000,None,50000]
})
print(df.isna())
```

```python
# 6 isna
# isna() returns True for missing and False for non-missing values.
result = df["Age"].isna()
print(result)
```

```python
# 7 isnull
# works the same way as isna()
result = df["Age"].isnull()
print(result)
```

```python
# 8 notna
#  returns True when a value is present and False when it is missing.
result = df["Age"].notna()
print(result)
```

```python
# 9 notnull
# works the same way as notna().
result = df["Age"].notnull()
print(result)
```

```python
# 10 Count Missing Values
#  sum() counts True as 1, so it counts missing values in each column.
print(df.isna().sum())
```

```python
# 11 Total Missing Values
# The first sum counts by column; the second sum adds all column counts.
Total_missing = df.isna().sum().sum()
print(Total_missing)
```

```python
# 12 Remove Missing Values — dropna()
df = pd.DataFrame({
    "Name":["jenish","Kriya","parth"],
    "Age":[22,None,30]
})
result = df.dropna()
print(result)
```

```python
# 13 dropna(how='all')
# how='all' removes a row only when every value in that row is missing.
df = pd.DataFrame({
    "Name": ["jenish", None, "Kriya"],
    "Age": [22, None, 30]
})

result = df.dropna(how="all")
print(result)
```

```python
# 14 Fill Missing Values - fillna
# replaces missing values with a value you provide.
df = pd.DataFrame({"Age":[22,None,30]})
df["Age"] = df["Age"].fillna(25)
print(df)
```

```python
# 15 Fill With Mean
# The mean is the average. Here, (30000 + 50000) / 2 = 40000.
df = pd.DataFrame({
    "Salary":[30000,None,50000]
})
mean_salary = df["Salary"].mean()
df["Salary"] = df["Salary"].fillna(mean_salary)
print(df)
```

```python
# 16 Fill with Median
# Median is the middle value after sorting and is less affected by extreme values.
df = pd.DataFrame({
    "Salary":[25000,None,75000]
})
median_salary = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(median_salary)
print(df)
```

```python
# 17 Fill with Mode
# Mode is the most frequently occurring value
df = pd.DataFrame({
    "City":["Pune","Mumbai",None,"Pune"]
})
mode_city = df["City"].mode()[0]
df["City"] = df["City"].fillna(mode_city)
print(df)
```

```python
# 18 Forward Fill
# Forward fill copies the previous available value forward.
df = pd.DataFrame({
    "Sales":[100,None,None,400]
})
df["Sales"] = df["Sales"].ffill()
print(df) 
```

```python
# 19 Backward Fill
# Backward fill copies the next available value backward.
df = pd.DataFrame({
    "Sales":[100,None,None,400]
})
df["Sales"] = df["Sales"].bfill()
print(df) 
```

```python
# 20 Interpolation
# Interpolation estimates missing values between known values.
df = pd.DataFrame({
    "Sales":[100,None,None,400]
})
df["Sales"] = df["Sales"].interpolate()
print(df) 
```

```python
# NaN / None / pd.NA -> Missing values
# isna() -> Detect missing
# isnull() -> Same as isna()
# notna() -> Detect available
# notnull() -> Same as notna()
# isna().sum() -> Count missing by column
# dropna() -> Remove missing data
# fillna() -> Replace missing data
# mean() -> Average
# median() -> Middle value
# mode() -> Most frequent value
# ffill() -> Forward fill
# bfill() -> Backward fill
# interpolate() -> Estimate between known values
```
