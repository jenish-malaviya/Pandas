```python
import pandas as pd
```

```python
# 1 Sum
df = pd.DataFrame({
    "Score":[70,80,90,100]
})
print(df["Score"].sum())
```

```python
# 2 Mean
print(df['Score'].mean())
```

```python
# 3 Median
print(df['Score'].median())
```

```python
# 4 Mode
df = pd.DataFrame({
    "Score":[70,80,80,100]
})
print(df["Score"].mode())
```

```python
# 5 min
print(df['Score'].min())
```

```python
# 6 max
print(df['Score'].max())
```

```python
# 7 std
print(df['Score'].std(), 2)
```

```python
# 8 var
print(df['Score'].var(), 2)
```

```python
# 9 Count
df = pd.DataFrame({
    "Score":[70,80,None,100]
})
print(df['Score'].count())
```

```python
# 10 quantile
df = pd.DataFrame({
    "Score":[70,80,90,100]
})
print(df["Score"].quantile(0.50))
```

```python
# 11 describe
df = pd.DataFrame({
    "Score":[70,80,90,100]
})
print(df["Score"].describe())
```

```python
# 12 cumsum
# cumsum() creates a running total.
df = pd.DataFrame({
    "Sales":[100,200,300,400]
})
print(df["Sales"].cumsum())
```

```python
# 13 cumprod
df = pd.DataFrame({"Value":[2,3,4,5]})
print(df["Value"].cumprod())
```

```python
# 14 cummin
# cummin() returns the smallest value seen so far
df = pd.DataFrame({"Sales":[500,300,400,200]})
print(df["Sales"].cummin())
```

```python
# 15 cummax
# cummax() returns the largest value seen so far.
df = pd.DataFrame({"Sales":[500,700,600,900]})
print(df["Sales"].cummax())
```

```python
# QUICK REVISION

# sum() -> Total
# mean() -> Average
# median() -> Middle value
# mode() -> Most frequent value
# min() -> Smallest
# max() -> Largest
# std() -> Standard deviation
# var() -> Variance
# count() -> Non-missing count
# quantile() -> Percentile value
# describe() -> Statistical summary
# cumsum() -> Running total
# cumprod() -> Running product
# cummin() -> Running minimum
# cummax() -> Running maximum
```
