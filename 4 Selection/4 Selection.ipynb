```python
import numpy as np 
import pandas as pd
```

```python
# Common DataFrame Used in All Examples
data = {
 "Name": ["Jenish", "Kriya", "Parth", "Aesha"],
 "Age": [22, 25, 21, 24],
 "City": ["Pune", "Mumbai", "Delhi", "Pune"],
 "Marks": [85, 92, 78, 88]
}
df = pd.DataFrame(data)
print(df)
```

```python
# 1 Selecting One Column
# selects only the Name column. The result is a Pandas Series.
name = df["Name"]
print(name)
```

```python
# 2 Selecting Multiple Columns
# Use double square brackets [[...]] when selecting more than one column.
result = df[["Name","Marks"]]
print(result)
```

```python
# 3  Selecting Rows
#  iloc[1] selects the second row because Pandas positions start at 0.
result = df.iloc[1]
print(result)
```

```python
# 4 Selecting Specific Cells
# iloc[1, 3] means row position 1 and column position 3. It returns the value 92
result = df.iloc[1 ,3]
print(result)
```

```python
# 5 loc 
# loc[] selects using row labels and column names. Here row label 1 and column Name are
# used.
result = df.loc[1,"Name"]
print(result)
```

```python
# 6 iloc 
# iloc[] selects using integer positions. Row 1 means the second row and column 0 means
# the first column.
result = df.iloc[1,0]
print(result)
```

```python
# 7  Difference Between loc and iloc
# Both return Delhi here. loc uses labels/names; iloc uses integer positions.
print(df.loc[2,"City"])
print(df.iloc[2,2])
```

```python
# 8 Selecting Rows by Index
#  loc[0:2] selects rows whose labels are 0 through 2. With loc, the ending label is included.
result = df.loc[0:2]
print(result)
```

```python
# 9 Selecting Columns by Name
# The colon means all rows, and "City" selects the City column.
result = df.loc[:,"City"]
print(result)
```

```python
# 10 Row + Column Selection
# The first part selects rows 1 to 2. The second part selects Name and Marks.
result = df.loc[1:2,["Name","Marks"]]
print(result)
```

```python
# 11 Slicing Rows
#  iloc[0:3] selects positions 0, 1, and 2. The ending position 3 is not included.
result = df.iloc[0:3]
print(result)
```

```python
# 12 Slicing Columns
# The colon means all rows. 1:4 selects columns at positions 1, 2, and 3.
result = df.iloc[:,1:4]
print(result)
```

```python
# Quick Revision — loc vs iloc

# loc = label-based selection
#  df.loc[1, "Name"] -> row label 1, column Name

# iloc = position-based selection
#  df.iloc[1, 0] -> second row, first column

# Important:
# loc -> uses labels and column names
# iloc -> uses integer positions
# loc slicing -> ending label is included
# iloc slicing -> ending position is excluded
```
