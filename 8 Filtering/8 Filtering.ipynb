```python
import pandas as pd
```

```python
# 1 Comparison Operators
#  Comparison operators compare values and return True or False.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John", "Neha"],
 "Age": [22, 25, 30, 20],
 "Salary": [30000, 45000, 60000, 25000]
})
print(df)
```

```python
# 2 Greater Than >
# Selects rows where Age is greater than 23
result = df[df["Age"] > 23]
print(result)
```

```python
# 3 Less Than <
# Selects rows where Age is less than 25
result = df[df["Age"] < 25]
print(result)
```

```python
# 4 Greater Than or Equal >=
# Includes 25 and values greater than 25
result = df[df["Age"] >= 25]
print(result)
```

```python
# 5  Less Than or Equal <=
#  Includes 22 and values below 22.
result = df[df["Age"] <= 22]
print(result)
```

```python
# 6 Equal ==
#  Selects rows where Age exactly equals 25.
result = df[df["Age"] == 25]
print(result)
```

```python
# 7  Not Equal !=
#  Selects rows where Age is not 25
result = df[df["Age"] != 25]
print(result)
```

```python
# 8 Boolean Conditions
# A Boolean condition gives True or False for every row.
condotion = df["Age"] > 23
print(condotion)
```

```python
# 9 AND &
# & means AND. Both conditions must be True.
result = df[(df["Age"] > 20) & (df["Salary"] > 30000)]
print(result)
```

```python
# 10 OR |
# | means OR. At least one condition must be True.
result = df[(df["Age"] < 21) | (df["Salary"] > 50000)]
print(result)
```

```python
# 11 NOT ~
# ~ reverses a Boolean condition.
result = df[~(df["Age"] > 23)]
print(result)
```

```python
# 12 Multiple Conditions
# Several conditions can be combined using & and |.
result = df[
    (df["Age"] >= 22) &
    (df["Salary"] >= 30000) &
    (df["Name"] != "John")
]
print(result)
```

```python
# 13 isin
# checks whether each value exists in a list.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","parth","Laxsh"]
})

result = df[df["Name"].isin(["Jenish","Kriya"])]
print(result)
```

```python
# 14 between
#  between(22, 25) selects values from 22 through 25; both ends are included by default.
df = pd.DataFrame({
    "Age":[25,41,23,22]
})

result = df[df["Age"].between(22,25)]
print(result)
```

```python
# 15 query
# lets you write a readable filtering expression as a string.
df = pd.DataFrame({
    "Age": [25, 41, 23, 22],
    "Salary": [25000, 12000, 85000, 100000]
})

result = df.query("Age >= 23 and Salary > 40000")
print(result)
```

```python
# 16 query() with OR
# Inside query(), use words such as and and or.
df = pd.DataFrame({
    "Age": [25, 41, 23, 22],
    "Salary": [25000, 12000, 85000, 100000]
})

result = df.query("Age < 21 or Salary > 50000")
print(result)
```

```python
# 17 Filter Text
# Text values can be filtered by comparing them with a string.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","parth","Laxsh"]
})
result = df[df["Name"] == "Kriya"]
print(result)
```

```python
# 18 Filter by Salary
# Selects employees whose salary is at least 40,000.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","parth","Laxsh"],
    "Salary": [25000, 12000, 85000, 100000]
})

result = df[df["Salary"] >= 40000]
print(result)
```
