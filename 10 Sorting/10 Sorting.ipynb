```python
import pandas as pd
```

```python
# 1  sort_values
# sorts rows using values in one or more columns.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth","Laxsh"],
    "Age":[22,30,25,20],
    "Salary":[30000,60000,45000,25000]
})
result = df.sort_values("Age")
print(result)
```

```python
# 2 Ascending Sorting
# ascending=True sorts from the smallest value to the largest. 
result = df.sort_values("Salary", ascending=True)
print(result)
```

```python
# 3  Descending Sorting 
# ascending=False sorts from largest to smallest.
result = df.sort_values("Salary", ascending=False)
print(result)
```

```python
# 4 Sort Multiple Columns
# The first column is the primary sort and the second column breaks ties.
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 50000, 30000],
    "Name": ["Amit", "Riya", "John", "Neha"]
})
result = df.sort_values(
    ["Department","Salary"],
    ascending=[True,False]
)
print(result)
```

```python
# 5 sort_index
# sorts the DataFrame using its index labels.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"]},
    index = [20,10,30]
)
result = df.sort_index()
print(result)
```

```python
# 6 Sorting by Index
# ascending=False sorts index labels from largest to smallest.
result = df.sort_index(ascending=False)
print(result)
```

```python
# 7 Missing Values While Sorting
#  By default, missing values are placed at the end when sorting.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John", "Neha"],
 "Salary": [30000, None, 45000, 25000]
})
result = df.sort_values("Salary")
print(result)
```

```python
# 8 Missing Values — NaN First
# na_position='first' puts missing values before non-missing values.
result = df.sort_values(
    "Salary",
    na_position="first"
)
print(result)
```

```python
# 9 Stable Sorting
# A stable sort preserves the original relative order of rows with equal values.
df = df = pd.DataFrame({
     "Name": ["Amit", "Riya", "John", "Neha"],
     "Score": [90, 90, 80, 90],
     "Order": [1, 2, 3, 4]
})
result = df.sort_values(
    "Score",
     kind="stable"
)
print(result)
```

```python
# sort_values("Age") -> Sort by column
# ascending=True -> Smallest to largest
# ascending=False -> Largest to smallest
# sort_values(["A","B"]) -> Sort multiple columns
# sort_index() -> Sort by index
# na_position="first" -> Missing values first
# na_position="last" -> Missing values last
# kind="stable" -> Preserve order of equal values
```
