```python
import pandas as pd
```

```python
# 1 what is group by 
# GroupBy divides data into groups, performs a calculation on each group, and combines the results.
# Think: Split → Apply → Combine.
df = pd.DataFrame({
     "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
     "Employee": ["A", "B", "C", "D", "E", "F"],
     "Salary": [50000, 60000, 45000, 55000, 40000, 50000]
})
print(df)
```

```python
# 2 group by 
# creates groups based on a column.
result = df.groupby("Department")["Salary"].sum()
print(result)
```

```python
# 3 Group by one column
# Group by Department and calculate the average salary.
result = df.groupby("Department")["Salary"].mean()
print(result)
```

```python
# 4 Group by multiple columns
# You can group using two or more columns.
df2 = pd.DataFrame({
 "Department": ["IT", "IT", "HR", "HR", "IT", "HR"],
 "Gender": ["M", "F", "F", "M", "M", "F"],
 "Salary": [50000, 60000, 45000, 55000, 70000, 50000]
})

result = df2.groupby(["Department","Gender"])["Salary"].mean()
print(result)
```

```python
# 5 sum
#  adds values inside each group.
result = df.groupby("Department")["Salary"].sum()
print(result)
```

```python
# 6 mean
# calculates the average of each group.
result = df.groupby("Department")["Salary"].mean()
print(result)
```

```python
# 7 count
# counts non-missing values in a column.
result = df.groupby("Department")["Salary"].count()
print(result)
```

```python
# 8 min
# finds the smallest value in each group.
result = df.groupby("Department")["Salary"].min()
print(result)
```

```python
# 9 max
# finds the largest value in each group.
result = df.groupby("Department")["Salary"].max()
print(result)
```

```python
# 10 median
# finds the middle value of each group.
result = df.groupby("Department")["Salary"].median()
print(result)
```

```python
# 11 std
# calculates the standard deviation for each group.
result = df.groupby("Department")["Salary"].std()
print(result)
```

```python
# 12 size
# counts the total number of rows in each group, including rows with missing values.
result = df.groupby("Department").size()
print(result)
```

```python
# 13 value_counts
# counts how many times each value appears.
result = df["Department"].value_counts()
print(result)
```

```python
# 14 agg
# lets you apply an aggregation function to grouped data.
result = df.groupby("Department")["Salary"].agg("mean")
print(result)
```

```python
# 15 Multiple aggregations
# Use a list when you want several calculations at the same time.
result = df.groupby("Department")["Salary"].agg(
    ["sum","mean","min","max","count"]
)
print(result)
```

```python
# 16 Named aggregation
# Named aggregation gives meaningful names to the result columns.
result = df.groupby("Department").agg(
    total_salary = ("Salary","sum"),
    average_salary = ("Salary","mean"),
    highest_salary = ("Salary","max")
)
print(result)
```

```python
# 17 transform
# returns a result with the same number of rows as the original DataFrame.
# Example: add each employee's department average.
df["Dept_Avg"] = df.groupby("Department")["Salary"].transform("mean")
print(df[["Department","Employee","Salary","Dept_Avg"]])
```

```python
# 18 filter
# keeps or removes complete groups based on a condition.
# Example: keep departments whose average salary is greater than 48000.
result = df.groupby("Department").filter(
    lambda groupby:groupby["Salary"].mean() > 48000
)
print(result)
```

```python
# 19 apply
# lets you run a custom function on each group.
def salary_range(group):
    return group["Salary"].max() - group["Salary"].min()

result = df.groupby("Department").apply(salary_range)
print(result)
```

```python
# 20 GroupBy with reset_index
# converts the grouping index back into a normal DataFrame column.
result = df.groupby("Department")["Salary"].sum().reset_index()
print(result)
```

```python
# 21 Multiple columns + multiple aggregations
result = df2.groupby("Department").agg(
 total_salary=("Salary", "sum"),
 average_salary=("Salary", "mean"),
 employee_count=("Salary", "count"),
 minimum_salary=("Salary", "min"),
 maximum_salary=("Salary", "max")
)
print(result)
```

```python
# 22 count() vs size()
df3 = pd.DataFrame({
 "Department": ["IT", "IT", "HR"],
 "Salary": [50000, None, 45000]
})
print(df3.groupby("Department")["Salary"].count())
print(df3.groupby("Department").size())
```

```python
# groupby() → creates groups
# sum() → total
# mean() → average
# count() → non-null count
# min() → smallest value
# max() → largest value
# median() → middle value
# std() → standard deviation
# size() → number of rows
# value_counts() → frequency of values
# agg() → multiple/custom aggregations
# transform() → result stays aligned with original rows
# filter() → keeps/removes complete groups
# apply() → custom function on each group
```
