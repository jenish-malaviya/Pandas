```python
import pandas as pd
```

```python
# 1 Find Dublicates
# Duplicates are repeated rows or repeated values in a DataFrame. First, create a DataFrame containing a
# repeated row.
df = pd.DataFrame({
    "Name":["A","B","B","C"],
    "Age":[20,25,25,30]
})
print(df)
```

```python
# 2 dublicated
# checks whether each row is a duplicate of an earlier row. It returns True for repeated rows and
# False for the first occurrence.
mask = df.duplicated()

print(mask)
print(df[mask])
```

```python
# 3 Remove Duplicate
# To remove duplicate rows, use `drop_duplicates()`. The first occurrence is kept by default.
result = df.drop_duplicates()
print(result)
```

```python
# 4 drop_duplicates
# returns a DataFrame after removing duplicate rows. It does not change the original
# DataFrame unless `inplace=True` is used.
result = df.drop_duplicates()

print("Original:")
print(df)

print("\n After removing duplicates:")
print(result)
```

```python
# 5 Duplicate Rows
# Here we create two completely identical rows. `duplicated()` identifies only the later occurrence by default.
df = pd.DataFrame([
    [10,20],
    [30,40],
    [10,20],
    [50,60]
],columns=["A","B"])

print(df)

print("\n Duplicate Rows:")
print(df[df.duplicated()])
```

```python
# 6 Duplicate Columns
# To find duplicate columns, transpose the DataFrame first. After `df.T`, original columns become rows. Then
# `duplicated()` can compare them.
df = pd.DataFrame([
    [10,20,10],
    [30,40,30],
    [50,60,50]
],columns=["A","B","C"])

print(df)

duplicate_columns = df.T.duplicated()

print("\n Duplicate column mask:")
print(duplicate_columns)

print("\n Duplicate column:")
print(df.T[duplicate_columns])
```

```python
# 7 Keep First
# `keep='first'` keeps the first occurrence and marks later duplicates as True. This is the default behavior.
df = pd.DataFrame({
    "Name":["A","B","B","B","C"],
    "Age":[20,25,25,25,30]
})
print(df.duplicated(keep="first"))
```

```python
# 8 Keep Last
# `keep='last'` keeps the last occurrence. Earlier occurrences are marked as duplicates.
print(df.duplicated(keep="last"))
```

```python
# 9 Remove all Duplicates
# To remove every row that belongs to a duplicated group, use `keep=False` with `duplicated()` and keep only
# rows that are not duplicates.
mask = df.duplicated(keep=False)

print("All duplicate rows:")
print(df[mask])

print("\nDataFrame after removing all duplicate groups:")
print(df[~mask])
```

```python
# Find duplicate rows           df.duplicated()                True for repeated rows after the first
# Remove duplicates             df.drop_duplicates()           Keeps first occurrence by default
# Keep first                    df.duplicated(keep='first')    Keeps first; marks later duplicates
# Keep last                     df.duplicated(keep='last')     Keeps last; marks earlier duplicates
# Find all duplicate members    df.duplicated(keep=False)      Marks every row in a duplicate group
# Remove all duplicate groups   df[~df.duplicated(keep=False)] Keeps only rows that occur once
# Find duplicate columns        df.T.duplicated()              Compares columns after transposing
```
