```python
import pandas as pd
```

```python
# 1 Why Combine DataFrames
# We can combine these two DataFrames into one DataFrame.
df1 = pd.DataFrame({
    "Name":["Jenish","Parth"],
    "Salary":[30000,40000]
})

df2 = pd.DataFrame({
    "Name":["Vidhi","Vishwa"],
    "Salary":[35000,45000]
})

print(df1)
print(df2)
```

```python
# 2 concat
# combines DataFrames. By default, it keeps the original indexes.
df1 = pd.DataFrame({
    "Name":["Jenish","Vidhi"],
    "Age":[22,23]
})

df2 = pd.DataFrame({
    "Name":["vishwa","swaru"],
    "Age":[22,23]
})

result = pd.concat([df1,df2])
print(result)
```

```python
# 3 Vertical Concatenation
# axis=0 means rows are added vertically.
result = pd.concat([df1,df2],axis=0)
print(result)
```

```python
# 4 Vertical Concatenation with ignore_index=True
# ignore_index=True creates a new continuous index.
result = pd.concat([df1,df2],axis=0,ignore_index=True)
print(result)
```

```python
# 5  Horizontal Concatenation
# axis=1 combines columns side by side using the index.
df1 = pd.DataFrame({
    "Name":["Jenish","Vidhi"]
})

df2 = pd.DataFrame({
    "Age":[21,23]
})

result = pd.concat([df1,df2], axis=1)
print(result)
```

```python
# 6 merge
# combines related tables using a common key.
employees = pd.DataFrame({
    "EmpID":[1,2,3],
    "Name":["jenish","vidhi","swara"]
})

salary = pd.DataFrame({
    "EmpID":[1,2,3],
    "Salary": [30000,40000,50000]
})

result = pd.merge(employees,salary,on="EmpID")
print(result)
```

```python
# 7 Inner Join 
# Inner join keeps only matching IDs from both DataFrames.
left = pd.DataFrame({
    "ID":[1,2,3],
    "Name":["jenish","parth","vidhi"]
})

right = pd.DataFrame({
    "ID":[2,3,4],
    "Salary":[40000,50000,60000]
})

result = pd.merge(left,right,on="ID",how="inner")
print(result)
```

```python
# 8 Left Join
# Left join keeps every row from the left DataFrame.
result = pd.merge(left,right,on="ID",how="left")
print(result)
```

```python
# 9 Right Join
# Right join keeps every row from the right DataFrame.
result = pd.merge(left,right,on="ID",how="right")
print(result)
```

```python
# 10 Outer Join
# Outer join keeps all IDs from both DataFrames.
result = pd.merge(left,right,on="ID",how="outer")
print(result)
```
