```python
import pandas as pd
```

```python
# 1 Add Column 
#  df['City'] creates a new column. Each value is placed in the matching row
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"],
    "Age":[22,24,23]
})
df["City"] = ["Pune","Mumbai","Delhi"]
print(df)
```

```python
# 2 Add multiple columns
# Two columns are added by making two assignments.
df["Salary"] = [30000,40000,35000]
df["Department"] = ["IT","HR","Finance"]
print(df)
```

```python
# 3 Add row
# combines DataFrames. ignore_index=True creates a new index.
new_row = pd.DataFrame({
    "Name":["Jenish"],
    "Age":[25],
    "City":["Surat"]
})
df = pd.concat([df,new_row],ignore_index=True)
print(df)
```

```python
# 4 assign
# adds one or more columns and returns a new DataFrame.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"],
    "Age":[22,24,23]
})
df = df.assign(
    City = ["Pune","Mumbai","Delhi"],
    Salary = [30000,40000,35000]
)
print(df)
```

```python
# 5 insert
#  insert(1, ...) puts Salary at position 1, so it becomes the second column.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"],
    "Age":[22,24,23],
    "City":["Pune","Mumbai","Delhi"]
})
df.insert(1,"Salary",[30000,40000,35000])
print(df)
```

```python
# 6 Remove Column using drop
# axis=1 means columns, so Age is removed.
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"],
    "Age":[22,24,23],
    "City":["Pune","Mumbai","Delhi"]
})
df = df.drop("Age",axis=1)
print(df)
```

```python
# 7 Remove Multiple Columns
#  Pass a list of column names to remove multiple columns.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John"],
 "Age": [22, 24, 23],
 "City": ["Pune", "Mumbai", "Delhi"]
})
df = df.drop(["Age","City"],axis=1)
print(df)
```

```python
# 8 Remove Row using drop
# drop(1) removes the row with index label 1. The index is not automatically reset.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John"],
 "Age": [22, 24, 23]
})
df = df.drop(1)
print(df)
```

```python
# 9 Remove Multiple Rows
#  Passing a list of index labels removes multiple rows.
df = df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John"],
 "Age": [22, 24, 23]
})
df = df.drop([0,2])
print(df)
```

```python
# 10 drop_duplicates
# removes duplicate rows and keeps the first occurrence by default.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "Amit", "John"],
 "Age": [22, 24, 22, 23]
})
df = df.drop_duplicates()
print(df)
```

```python
# 11 Duplicates using subset
# subset=['Name'] checks duplicates using only Name.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "Amit", "John"],
 "Age": [22, 24, 25, 23]
})

df = df.drop_duplicates(subset=["Name"])
print(df)
```

```python
# 12 Rename Columns using rename
#  rename(columns={old:new}) changes selected column labels.
df = pd.DataFrame({
    "Name":["Jenish","Kriya"],
    "Age":[22,24]
})
df = df.rename(columns={
    "Name":"Student_Name",
    "Age":"Student_Age"
})
print(df)
```

```python
# 13 Rename Index using rename
# rename(index={old:new}) changes index labels.
df = pd.DataFrame({
    "Name":["Jenish","Kriya"],
    "Age":[22,24]
})
df = df.rename(index={
    0:"Student_1",
    1:"Student_2"
})
print(df)
```

```python
# 14 Rename All Columns
# df.columns replaces all column names. The number of names must match the number of columns.
df = pd.DataFrame({
 "Name": ["Amit", "Riya"],
 "Age": [22, 24]
})
df.columns = ["Student_Name","Student_Age"]
print(df)
```
