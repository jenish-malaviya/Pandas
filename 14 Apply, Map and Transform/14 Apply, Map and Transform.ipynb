```python
import pandas as pd
```

```python
# 1 map
# map() applies a function to every item of an iterable. It returns a map object, so list() is often used to display the result.
numbers = [1,2,3,4]
result = list(map(lambda x: x * 2,numbers))
print(result)
```

```python
# 2 apply with a series
# Series.apply() calls the function for each value in the Series.
s = pd.Series([10,20,30])
result = s.apply(lambda x: x + 5)
print(result)
```

```python
# 3 DataFrame element-wise operations
# For simple arithmetic, Pandas/NumPy vectorized operations are preferred because they are concise and generally
# faster than Python-level function calls
df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})
result = df * 10
print(result)
```

```python
# 4 DataFrame.map() / older applymap()
df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})
result = df.map(lambda x: x * 2)
print(result) 
```

```python
# 5 tranform
# transform() returns a result aligned to the original shape. This is particularly valuable with grouped data.
s = pd.DataFrame([10,20,30])
result = s.transform(lambda x: x + 100)
print(result)
```

```python
# 6 lambda functions
# A lambda is a small anonymous function. Syntax: lambda arguments: expression.
square = lambda x: x ** 2
print(square(5))
```

```python
# 7 Functions application to a series
# A named custom function can be passed directly to Series.apply().
def add_tax(x):
    return x * 1.18

prices = pd.Series([100,200,300])
result = prices.apply(add_tax)
print(result)
```

```python
# 8 Function application to a DataFrame
# By default, DataFrame.apply() works column-by-column (axis=0).
df = pd.DataFrame({
    "A": [10,20],
    "B": [30,40]
})

def total(column):
    return column.sum()
    
result = df.apply(total)
print(result)
```

```python
# 9 DataFrame apply() row-wise
# axis=1 passes each row to the function. axis=0 (default) passes each column.
df = pd.DataFrame({
 "A": [10, 20],
 "B": [30, 40]
})

result = df.apply(lambda row: row["A"] + row["B"], axis=1)
print(result)
```

```python
# 10 Custom function
# Custom functions are useful when the transformation needs multiple steps or conditions.
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    else:
        return "C"

marks = pd.Series([95,82,65])
result = marks.apply(grade)
print(result)
```

```python
# 11 GroupBy + transform()
df = pd.DataFrame({
 "Department": ["IT", "IT", "HR", "HR"],
 "Salary": [50000, 70000, 40000, 60000]
})

df["Dept_Avg"] = df.groupby("Department")["Salary"].transform("mean")
print(df)
```
