```python
import pandas as pd
```

```python
# 1 What is Index
# Index is the label used to identify each row in a Pandas Series or DataFrame.
# Think of an index like a row number or ID. It helps Pandas locate and select rows.
df = pd.DataFrame({
    "Name": ["Jenish","Kriya","Parth"],
    "Age":[21,22,23]
})
print(df)
```

```python
# 2 Default Index
# When you create a DataFrame without specifying an index, Pandas automatically creates a default integer
# index starting from 0.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John"],
 "Marks": [80, 90, 85]
})
print(df.index)
print(df)
```

```python
# 3 Custom Index
# You can give your own labels to rows instead of using 0, 1, 2.
df = pd.DataFrame(
    {
        "Name":["jenish","kriya","parth"],
        "Marks":[80,90,85]
    },
    index=["s1","s2","s3"]
)
print(df)
```

```python
# 4 Changing Index
# You can replace the existing index with a new list of labels. The number of new labels must match the
# number of rows.
df = pd.DataFrame({
    "Name":["jenish","Kriya","parth"],
    "Marks":[80,90,85]
})
df.index = ["s1","s2","s3"]
print(df)
```

```python
# 5 Set_index
# makes an existing column the DataFrame index. By default, that column is removed from the
# normal columns.
df = pd.DataFrame({
    "ID":[101,102,103],
    "Name":["Jenish","Kriya","Parth"],
    "Marks":[80,90,85]
})

df = df.set_index("ID")
print(df)
```

```python
# 6 set_index() with inplace=True
# inplace=True changes the original DataFrame directly. Without inplace=True, set_index() returns a new
# DataFrame.
df = pd.DataFrame({
    "Id":[101,102,103],
    "Name":["Jneish","Kriya","Parth"]
})

df.set_index("Id",inplace=True)
print(df)
```

```python
# 7  reset_index
# changes the current index back into a normal column and creates a new default index 0, 1, 2,
df = pd.DataFrame({
    "id":[101,102,103],
    "Name":["jenish","Kriya","Parth"]
})

df = df.set_index("id")
df = df.reset_index()
print(df)
```

```python
# 8 reset_index(drop=True)
# drop=True removes the old index instead of converting it into a column. This is useful when you only want a
# fresh 0-based index.
df = pd.DataFrame({
    "Name":["jenish","kriya","parth"]
},index=["A","B","C"])
df = df.reset_index(drop=True)
print(df)
```

```python
# 9 Index Selection
# You can select rows using their index labels with loc[].
df = pd.DataFrame({
    "Name":["Jenish","Kriya","Parth"],
    "Marks":[80,90,85]
},index=["S1","S2","S3"])
print(df.loc["S2"])
```

```python
# 10 Select Multiple Index Labels
df = pd.DataFrame({
    "Name":["Jenish","Jemin","Nity"],
    "Marks":[80,45,22]
}, index=["S1","S2","S3"])
print(df.loc[["S1","S2"]])
```

```python
# 11 Index Slicing
# Index slicing selects a range of rows. With label-based slicing using loc[], both the start and end labels are
# included.
df = pd.DataFrame({
    "Name":["Jenish","Nity","Vidhi","Parth","Sneh"],
    "Marks":[52,45,96,62,41]
},index=["J1","J2","J3","J4","J5"])
print(df.loc["J2":"J4"])
```

```python
# 12 Index Slicing with iloc[]
# iloc[] selects rows by integer position. Here, positions start from 0. The stop position is excluded, just like
# normal Python slicing.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John", "Neha"],
 "Marks": [80, 90, 85, 95]
}, index=["S1", "S2", "S3", "S4"])
print(df.iloc[1:3])
```

```python
# 13 Duplicate Indexes
# Pandas allows duplicate index labels. However, selecting a duplicated label can return multiple rows.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John"],
 "Marks": [80, 90, 85]
}, index=["S1", "S1", "S2"])
print(df)
print()
print(df.loc["S1"])
```

```python
# 14 Check for Duplicate Indexes
# index.duplicated() tells you which index labels are duplicates. True means that the label has appeared
# before.
df = pd.DataFrame({
 "Name": ["Amit", "Riya", "John"]
}, index=["S1", "S1", "S2"])
print(df.index.duplicated())
```

```python
# 15 MultiIndex Introduction
# A MultiIndex is an index with more than one level. It is useful when data has multiple grouping levels, such
# as Department and Student.

df = pd.DataFrame({
    "Marks":[80,56,95,85]
},index=[
    ["IT", "IT", "HR", "HR"],
    ["Amit", "Riya", "John", "Neha"]
])
df.index.names = ["Department","Student"]
print(df)
```

```python
# 16  Creating MultiIndex with from_tuples
# MultiIndex.from_tuples() creates a MultiIndex from pairs or tuples.
index = pd.MultiIndex.from_tuples(
    [
        ("IT","Jenish"),
        ("IT","Kriya"),
        ("HR","Parth")
    ],
    names=["Department","Student"]
)
df = pd.DataFrame({"Marks":[80,52,45]},index=index)
print(df)
```

```python
# 17 Selecting from MultiIndex
# loc[] can select data using one level or multiple levels of a MultiIndex.
index = pd.MultiIndex.from_tuples(
    [("IT","Jenish"),("IT","Kriya"),("HR","Parth")],
    names = ["Department","Student"]
)
df = pd.DataFrame({"Marks":[80,90,85]},index=index)
print(df.loc["IT"])
```

```python
# 18 Selecting One Exact MultiIndex Row
# Give both index labels as a tuple to select one exact row.
index = pd.MultiIndex.from_tuples(
    [("IT","Jenish"),
     ("IT","Kriya"),
     ("HR","Parth")
    ],
    names = ["Department","Student"]
)
df = pd.DataFrame({"Marks":[80,90,85]},index=index)
print(df.loc[("IT","Jenish")])
```

```python
# 19 Interview Difference: loc vs iloc
# loc[] = label-based selection. iloc[] = position-based selection.
# Example: If the index is ["S1", "S2", "S3"]: df.loc["S2"] selects the row labeled S2. df.iloc[1] selects the
# second row by position.

df = pd.DataFrame(
 {"Name": ["Amit", "Riya", "John"]},
 index=["S1", "S2", "S3"]
)
print(df.loc["S2"])
print()
print(df.iloc[1])
```
