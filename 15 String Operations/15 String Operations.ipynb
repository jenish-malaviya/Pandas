```python
import pandas as pd
```

```python
# Example
names = pd.Series(["Jenish Malaviya",
                   "Vidhi Malaviya",
                   "Vishwa Malaviya",
                   "Swara Malaviya"])
print(names)
```

```python
# 1 String accessor
# The .str accessor tells Pandas that we want to perform a string operation on each value of the Series.
s = pd.Series(["Python","Pandas","Numpy"])
print(s.str.lower())
```

```python
# 2 lower
# Converts letters to lowercase.
s = pd.Series(["HELLO","python","PANDAS"])

result = s.str.lower()
print(result)
```

```python
# 3 upper
# Converts letters to uppercase.
s = pd.Series(["hello","Hello","pandas"])

result = s.str.upper()
print(result)
```

```python
# 4 title
# Capitalizes the first letter of each word.
s = pd.Series(["helo world",
              "python pandas",
              "data science"])

result = s.str.title()
print(result)
```

```python
# 5 capitilize 
# Capitalizes the first character and makes the remaining characters lowercase
s = pd.Series(["hello world","PYTHON","data SCIENCE"])

result = s.str.capitalize()
print(result)
```

```python
# 6 strip
# Removes whitespace from both the beginning and end.
s = pd.Series(["   hello    ","  python  ","pandas"])

result = s.str.strip()
print(result)
```

```python
# 7 lstrip
# Removes whitespace from the left side only.
s = pd.Series(["   hello","   python","pandas"])

result = s.str.lstrip()
print(result)
```

```python
# 8 rstrip
# Removes whitespace from the right side only.
s = pd.Series(["   hello","   python","pandas"])

result = s.str.rstrip()
print(result)
```

```python
# 9 replace
# Replaces matching text. With regex=False, the search is treated as ordinary text.
s = pd.Series(["I like Python","Python is easy","Learn Python"])

result = s.str.replace("Python","Pandas",regex=False)
print(result)
```

```python
# 10 contains
# Checks whether the text contains the specified pattern and returns True or False.
s = pd.Series(["Python","Pandas","Numpy","Python Pandas"])

result = s.str.contains("Python",na=False)
print(result)
```

```python
# 11 startswith
# Checks whether each string starts with the specified text.
s = pd.Series(["Python","Pandas","Numpy","Python Basic"])

result = s.str.startswith("Py")
print(result)
```

```python
# 12 endswith
# Checks whether each string ends with the specified text.
s = pd.Series(["hello.py","data.csv","notes.txt","model.py"])

result = s.str.endswith(".py")
print(result)
```

```python
# 13 find
# Returns the position of the first occurrence. If the text is not found, it returns -1. Positions start at 0.
s = pd.Series(["Python","Pandas","Data Science"])

result = s.str.find("a")
print(result)
```

```python
# 14 len
# Returns the number of characters in each string, including spaces.
s = pd.Series(["Python","Pandas","Data Science"])

result = s.str.len()
print(result)
```

```python
# 15 split
# Splits each string into a list using the specified separator.
a = pd.Series(["Alice Smith","Bob Jones","Charlie Brown"])

result = a.str.split(" ")
print(result)
```

```python
# 16 join
# Joins the elements of each list using the separator.
s = pd.Series([["Python","Pandas"],["Data","Science"]])

result = s.str.join(" - ")
print(result)
```

```python
# 17 extract
# Extracts groups from text using a regular expression. Parentheses create capture groups.
s = pd.Series([
    "Name: Jenish, Age:25",
    "Name: Parth, Age:30"
])

result = s.str.extract(r"Name: (\w+), Age:(\d+)")
print(result)
```

```python
# 18 Regular expressions with Pandas
# Regular expressions (regex) let you describe text patterns. Here, the pattern checks for a basic email-like structure
s = pd.Series([
 "alice@gmail.com",
 "bob@yahoo.com",
 "charlie@gmail.com",
 "invalid-email"
])
result = s.str.contains(r"^[\w.-]+@[\w.-]+\.\w+$", regex=True, na=False)
print(result)
```
