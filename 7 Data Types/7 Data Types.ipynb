```python
import pandas as pd
```

```python
# 1 dtype
# dtype means data type. It tells pandas what type of data is stored in a column.
df = pd.DataFrame({
    "Age":[22,24,23],
    "Salary":[30000.5,40000.5,35000.5],
    "Name":["Jenish","Kriya","Parth"]
})
print(df.dtypes)
```

```python
# 2 integer
# Integer contains whole numbers without decimal values, such as 10, 25, and 100.
df = pd.DataFrame({"Age":[22,24,23]})

print(df["Age"])
print(df["Age"].dtype)
```

```python
# 3 Float
# Float contains numbers with decimal values, such as 10.5 and 99.99.
df = pd.DataFrame({"Salary": [30000.5, 40000.75, 35000.25]})

print(df["Salary"])
print(df["Salary"].dtype)
```

```python
# 4 String
# Text data is commonly stored as object in pandas DataFrames. Pandas also has a dedicated string
# dtype.
df = pd.DataFrame({"Name": ["Amit", "Riya", "John"]})

print(df["Name"])
print(df["Name"].dtype)
```

```python
# 5 Boolean
# Boolean has two logical values: True and False.
df = pd.DataFrame({"Passed": [True, False, True]})

print(df["Passed"])
print(df["Passed"].dtype)
```

```python
# 6 Object 
# object is a general-purpose pandas dtype. It is commonly seen with text and can also hold mixed
# Python objects.
df = pd.DataFrame({"Data":["Jenish",25,"Pune"]})

print(df["Data"])
print(df["Data"].dtype)
```

```python
# 7 Category
#  Category is useful when a column contains a limited number of repeated values.
df = pd.DataFrame({
    "Department":["IT","HR","IT","Finance"]
})

df["Department"] = df["Department"].astype("category")
print(df["Department"])
print(df["Department"].astype)
```

```python
# 8 Datetime
# Datetime stores dates and times in a format pandas can analyze.
df = pd.DataFrame({
    "Date":["2026-01-10","2026-02-15","2026-03-20"]
})
df["Date"] = pd.to_datetime(df["Date"])
print(df["Date"])
print(df["Date"].dtype)
```

```python
# 9 dtypes
# dtypes shows the data type of every column in a DataFrame.
df = pd.DataFrame({
    "Age":[22,24,23],
    "Salary":[30000.5,40000.5,35000.5],
    "Passed":[True,False,True],
    "Name":["Jenish","Kriya","Parth"]
})
print(df.dtypes)
```

```python
# 10 astype
# converts a column to a specified dtype. Here, text numbers become integers.
df = pd.DataFrame({"Age":["22","24","23"]})
print(df.dtypes)
df["Age"] = df["Age"].astype(int)
print(df.dtypes)
print(df)
```

```python
# 11 to_numeric
# converts values into numeric data.
df = pd.DataFrame({"Salary":["30000","40000","35000"]})
df["Salary"] = pd.to_numeric(df["Salary"])

print(df)
print(df["Salary"].dtype)
```

```python
# 12  to_numeric() with errors='coerce'
# errors='coerce' converts invalid values to NaN instead of raising an error.
df = pd.DataFrame({"Marks":["90","85","ABC"]})
df["Marks"] = pd.to_numeric(df["Marks"],errors = "coerce")
print(df)
```

```python
# 13 convert_dtypes
# tries to convert columns to suitable pandas nullable dtypes.
df = pd.DataFrame({
    "Age":[22,24,23],
    "Name":["Jenish","Kriya","Parth"],
    "Passed":[True,False,True]
})
df = df.convert_dtypes()
print(df.dtypes)
```

```python
# 14 astype() vs to_numeric() vs convert_dtypes()
# These methods solve different conversion problems.

# # Convert to a specific type
# df["Age"] = df["Age"].astype(int)
# # Convert values to numeric
# df["Salary"] = pd.to_numeric(df["Salary"])
# # Automatically choose suitable pandas dtypes
# df = df.convert_dtypes()
```

```python
# 15 Complete Data Types Example

df = pd.DataFrame({
 "Age": [22, 24, 23],
 "Salary": [30000.5, 40000.0, 35000.5],
 "Name": ["Amit", "Riya", "John"],
 "Passed": [True, False, True],
 "Department": ["IT", "HR", "IT"],
 "JoiningDate": ["2026-01-10", "2026-02-15", "2026-03-20"]
})
df["Department"] = df["Department"].astype("category")
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
print(df.dtypes)
```

```python
# dtype -> Data type
# int -> Whole numbers
# float -> Decimal numbers
# string/object -> Text
# bool -> True / False
# category -> Repeated categorical values
# datetime64[ns] -> Date and time
# dtypes -> Shows every column's dtype
# astype() -> Convert to a specified dtype
# to_numeric() -> Convert values to numeric
# convert_dtypes() -> Choose suitable pandas dtypes
```
