```python
# 1 CSV — pd.read_csv
# CSV means Comma-Separated Values. It is one of the most common data-file formats
df = pd.read_csv("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv")
print(df)
```

```python
# 2 Excel — pd.read_excel
df = pd.read_excel("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.xlsx")
print(df)
```

```python
# 3 JSON — pd.read_json
df = pd.read_json("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.json")
print(df)
```

```python
# 4 HTML — pd.read_html
# extracts HTML tables and normally returns a list of DataFrames.
tables = pd.read_html("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.html")
df = tables[0]
print(df)
```

```python
# 5 File Paths

# Absolute path using double backslashes
df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv"
)

print(df)

# Absolute path using a raw string
df = pd.read_csv(
    r"D:\Education\computer course\Pandas\My Notes\19 Read files\Pandas_Level_19_Practice_Files\employees.csv"
)

print(df)
```

```python
# 6 Delimiters — sep
# A delimiter separates values. Comma is common, but files may use semicolons or tabs.
df = pd.read_csv("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv", sep=";")
print(df)
```

```python
# 7 Headers - header and names
# Use the first row as the header:
df = pd.read_csv("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv")

# If the file has no header:
df = pd.read_csv("D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv")

print(df)
# Give your own column names:

df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
    header=None,
    names=["Name", "Age", "Sales"]
)

print(df)
```

```python
# 8 usecols
# usecols reads only selected columns.
df = pd.read_csv(
      "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
       usecols=["Name", "Department"]
)
print(df)
```

```python
# 9 skiprows
# skiprows skips rows while reading.
df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
     skiprows=3
)
print(df)

# Specific rows can also be skipped:
df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
     skiprows=[2,5])
print(df)
```

```python
# 10 nrows
# nrows limits how many rows are read.
df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
     nrows=2)
print(df)
```

```python
# 11 index_col
# index_col makes a selected column the DataFrame index.
# Which column you cant want to seen in table
df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
    index_col="City")
print(df)
```

```python
# 12 dtype
# dtype specifies the data type while reading.
df = pd.read_csv(
    "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
     dtype={"Age":"int64","Sales":"float64"}
)
print(df.dtypes)
```

```python
# 13 parse_dates
# parse_dates converts selected columns into datetime values.
df = pd.read_csv(
     "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
      parse_dates=["JoinDate"]
)
print(df.dtypes)
```

```python
# 14 Encoding
# Encoding controls how text bytes are decoded. UTF-8 is common.
df = pd.read_csv(
 "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees_utf8.csv",
 encoding="utf-8"
)
print(df)

df = pd.read_csv(
  "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees_utf8.csv",
   encoding="latin1"
)
print(df)
```

```python
# 15 Combine Parameters
df = pd.read_csv(
     "D:\\Education\\computer course\\Pandas\\My Notes\\19 Read files\\Pandas_Level_19_Practice_Files\\employees.csv",
     sep=",",
     header=0,
     usecols=["JoinDate", "Name", "Department"],
     nrows=5,
     dtype={"Name": "string"},
     parse_dates=["JoinDate"],
     encoding="utf-8"
)
print(df)
print(df.dtypes)
```

```python
# # Quick Revision

# CSV -> pd.read_csv()
# Excel -> pd.read_excel()
# JSON -> pd.read_json()
# HTML -> pd.read_html()

# Relative path -> from current folder
# Absolute path -> complete location

# sep -> delimiter
# header -> header row
# usecols -> selected columns
# skiprows -> skipped rows
# nrows -> number of rows to read
# index_col -> index column
# dtype -> data type
# parse_dates -> datetime conversion
# encoding -> text decoding
```
